```python
from google.colab import files
import pandas as pd

uploaded = files.upload()

filename = list(uploaded.keys())[0]

df = pd.read_csv(filename)

print("Dataset loaded successfully!")
display(df.head(10))
```

```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

count_vectorizer = CountVectorizer()

tf_matrix = count_vectorizer.fit_transform(
    df['Clean_Document']
)

tf_table = pd.DataFrame(
    tf_matrix.toarray(),
    columns=count_vectorizer.get_feature_names_out(),
    index=df['Document_ID']
)

print("TERM FREQUENCY (TF) TABLE")

display(tf_table.head(10))
```

```python
import numpy as np

df_values = (tf_matrix > 0).sum(axis=0).A1

df_table = pd.DataFrame({
    'Term': count_vectorizer.get_feature_names_out(),
    'DF': df_values
})

print("DOCUMENT FREQUENCY TABLE")

display(df_table.head(20))
```

```python
N = len(df)

idf_values = np.log(N / df_values)

idf_table = pd.DataFrame({
    'Term': count_vectorizer.get_feature_names_out(),
    'DF': df_values,
    'IDF': idf_values
})

print("INVERSE DOCUMENT FREQUENCY TABLE")

display(idf_table.head(20))
```

```python
tfidf_values = tf_matrix.toarray() * idf_values

tfidf_table = pd.DataFrame(
    tfidf_values,
    columns=count_vectorizer.get_feature_names_out(),
    index=df['Document_ID']
)

print("TF-IDF MATRIX")

display(tfidf_table.head(10))
```

```python
# Select the first term in the vocabulary
term = count_vectorizer.get_feature_names_out()[0]

# Select the first document
doc_id = df['Document_ID'].iloc[0]

TF = tf_table.loc[doc_id, term]
DF = df_table.set_index('Term').loc[term, 'DF']
IDF = np.log(N / DF)
TFIDF = TF * IDF

print("Document:", doc_id)
print("Term:", term)
print("TF:", TF)
print("DF:", DF)
print("IDF:", round(IDF, 4))
print("TF-IDF:", round(TFIDF, 4))

print(
    "Program verification:",
    round(tfidf_table.loc[doc_id, term], 4)
)
```

```python
# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Remove records without department names
df = df.dropna(
    subset=['Ministry/Department']
).copy()

# Create text documents
def create_document(row):

    department = str(row['Ministry/Department'])

    received = row['Received']
    disposed = row['Disposed']

    text = (
        f"{department} received {received} "
        f"public grievances and disposed "
        f"{disposed} grievances. "
        f"Government public grievance "
        f"redressal department."
    )

    return text

df['Document'] = df.apply(
    create_document,
    axis=1
)

# Assign document IDs
df['Document_ID'] = [
    f"D{i+1}" for i in range(len(df))
]

print("Total Documents:", len(df))

display(
    df[['Document_ID', 'Document']].head(10)
)
```

```python
import nltk
import re

from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

stemmer = PorterStemmer()

def preprocess_text(text):

    # Convert to lowercase
    text = text.lower()

    # Remove numbers and special characters
    text = re.sub(r'[^a-zA-Z\s]', '', text)

    # Tokenization
    tokens = text.split()

    # Stop-word removal
    tokens = [
        word for word in tokens
        if word not in stop_words
    ]

    # Stemming
    tokens = [
        stemmer.stem(word)
        for word in tokens
    ]

    return ' '.join(tokens)


df['Clean_Document'] = df['Document'].apply(
    preprocess_text
)

print("Text preprocessing completed!")

display(
    df[
        ['Document_ID', 'Document', 'Clean_Document']
    ].head(10)
)
```
